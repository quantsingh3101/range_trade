```python
import pandas as pd
import datetime
```

```python
mcx_instrument_df = pd.read_csv('mcx_instrument_df.csv')  # Load your MCX instrument data into a DataFrame
def get_mcx_futures_symbol(name: str) -> str:
    # Get all futures contracts for the given name
    contracts = mcx_instrument_df[(mcx_instrument_df['name'] == name) & \
                                mcx_instrument_df['tradingsymbol'].str.contains('FUT')][['tradingsymbol', 'expiry']].to_dict('records')
    # Sort contracts by expiry date (assuming earliest expiry first)
    contracts.sort(key=lambda x: datetime.datetime.strptime(x['expiry'], '%Y-%m-%d').date())
    today = datetime.datetime.now().date()
    # Find the first contract that is not close to expiry (more than 10 days away)
    for contract in contracts:
        expiry_date = datetime.datetime.strptime(contract['expiry'], '%Y-%m-%d').date()
        days_to_expiry = (expiry_date - today).days
        # If more than 10 days to expiry, return this contract
        if days_to_expiry > 10:
            return contract['tradingsymbol']
        

get_mcx_futures_symbol('SILVER')
```
